# PDF to Markdown Converter

Convert multiple PDF files to Markdown using Microsoft's [MarkItDown](https://github.com/microsoft/markitdown).

### Features
- Upload multiple PDF files at once
- Convert all PDFs to Markdown
- Keep the original filenames
- Create a ZIP file containing all Markdown files
- Download the ZIP directly from Google Colab

## Upload Multiple PDF Files

Run the cell below and select multiple PDF files from your computer.

````python
from google.colab import files

uploaded = files.upload()

pdf_files = [
    filename
    for filename in uploaded.keys()
    if filename.lower().endswith('.pdf')
]

print(f"Uploaded {len(pdf_files)} PDF file(s):")

for pdf_file in pdf_files:
    print(f"  - {pdf_file}")
````

## Convert PDFs to Markdown

````python
from markitdown import MarkItDown
import os

md = MarkItDown()
md_files = []

for pdf_file in pdf_files:
    print(f"Converting: {pdf_file}")

    try:
        result = md.convert(pdf_file)

        output_file = os.path.splitext(pdf_file)[0] + '.md'

        with open(output_file, 'w', encoding='utf-8') as f:
            f.write(result.text_content)

        md_files.append(output_file)
        print(f"✓ Saved: {output_file}\n")

    except Exception as e:
        print(f"✗ Failed: {pdf_file}")
        print(f"  Error: {e}\n")

print(f"Converted {len(md_files)} PDF file(s) successfully.")
````

## Preview the Converted Markdown

````python
if md_files:
    first_file = md_files[0]

    print(f"Preview: {first_file}\n")
    print('-' * 80)

    with open(first_file, 'r', encoding='utf-8') as f:
        preview = f.read()

    print(preview[:5000])
else:
    print('No Markdown files were created.')
````

## Create ZIP File

````python
import zipfile

zip_name = 'converted_markdown.zip'

with zipfile.ZipFile(zip_name, 'w', zipfile.ZIP_DEFLATED) as zipf:
    for md_file in md_files:
        zipf.write(md_file)

print(f"✓ Created: {zip_name}")
print(f"✓ Contains {len(md_files)} Markdown file(s)")
````

## Download the ZIP

````python
from google.colab import files

files.download(zip_name)
````
