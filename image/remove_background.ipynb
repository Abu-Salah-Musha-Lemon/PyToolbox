# 🖼️ Remove Image Background

Remove the background from one or multiple images using **rembg**.

### Features
- Upload multiple images at once
- Automatically remove backgrounds
- Preserve transparent backgrounds as PNG
- Download all results as a ZIP file
- Runs directly in Google Colab

## 2. Upload Images

You can select multiple images at the same time.

````python
from google.colab import files

uploaded = files.upload()

image_files = [
    filename
    for filename in uploaded.keys()
    if filename.lower().endswith(
        ('.png', '.jpg', '.jpeg', '.webp', '.bmp')
    )
]

print(f"Uploaded {len(image_files)} image(s):")

for image_file in image_files:
    print(f"  - {image_file}")
````

## 3. Remove Backgrounds

````python
from rembg import remove
from PIL import Image
import os

output_files = []

for image_file in image_files:
    print(f"Processing: {image_file}")

    input_image = Image.open(image_file).convert('RGBA')
    output_image = remove(input_image)

    base_name = os.path.splitext(image_file)[0]
    output_file = f"{base_name}_no_bg.png"

    output_image.save(output_file, 'PNG')
    output_files.append(output_file)

    print(f"✓ Saved: {output_file}\n")

print(f"Processed {len(output_files)} image(s) successfully.")
````

## 4. Preview Results

````python
from IPython.display import display

for output_file in output_files:
    print(output_file)
    display(Image.open(output_file))
````

## 5. Create ZIP File

````python
import zipfile

zip_name = 'images_without_background.zip'

with zipfile.ZipFile(zip_name, 'w', zipfile.ZIP_DEFLATED) as zipf:
    for output_file in output_files:
        zipf.write(output_file)

print(f"✓ Created: {zip_name}")
````

## 6. Download Results

````python
from google.colab import files

files.download(zip_name)
````
